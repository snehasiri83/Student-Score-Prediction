#1.import libraries and datasets


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("/content/student.csv")
x=dataset.iloc[:,:-1].values
y=dataset.iloc[:,-1].values
```

```python
dataset.shape
```

```python
dataset.info()
```

```python
dataset.head()
```

#2.dataset visulatization

```python

plt.scatter(x,y,color='blue')
plt.xlabel('hours')
plt.ylabel('score')
plt.title('exam score prediction')
plt.show()
```

#3.splitting Dataset


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

#4.fitting regression model

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

#5.predicting test results

```python
y_predict=model.predict(x_test)
print(y_predict)
```

#6.finding metrics for the dataset

```python
from sklearn.metrics import r2_score,mean_squared_error,root_mean_squared_error,mean_absolute_error
print("r2_score:",r2_score(y_test,y_predict))
print("MSE:",mean_squared_error(y_test,y_predict))
print("RMSE:",root_mean_squared_error(y_test,y_predict))
print("MAE:",mean_absolute_error(y_test,y_predict))
```

#7.visualizing the results

```python
plt.scatter(x_train,y_train,color='blue')
plt.plot(x_train,model.predict(x_train),color='red')
plt.xlabel('hours')
plt.ylabel('score')
plt.title('exam score prediction')
plt.show()
```
